```python
import cv2 
import numpy as np 
import matplotlib.pyplot as plt 
# Step 1: Load the image 
image =  cv2.imread('D:\\anaconda\\images\\img4.jpg', cv2.IMREAD_GRAYSCALE)
# Step 2: Apply Gaussian Blur to reduce noise 
gaussian_blur = cv2.GaussianBlur(image, (5, 5), 1.4) 
# Step 3: Compute the gradient magnitude and direction using Sobel operator 
# Compute gradients along the x and y axis 
grad_x = cv2.Sobel(gaussian_blur, cv2.CV_64F, 1, 0, ksize=3) 
grad_y = cv2.Sobel(gaussian_blur, cv2.CV_64F, 0, 1, ksize=3) 
# Compute gradient magnitude and direction 
gradient_magnitude = np.sqrt(grad_x**2 + grad_y**2) 
gradient_direction = np.arctan2(grad_y, grad_x) * (180 / np.pi) % 180  # In degrees 
# Normalize the gradient magnitude to 0-255 for visualization 
gradient_magnitude = cv2.convertScaleAbs(gradient_magnitude) 
# Step 4: Apply Canny Edge Detection for final edges 
edges = cv2.Canny(gaussian_blur, threshold1=100, threshold2=200) 
# Visualization with annotations 
# Create a figure to display all the images in different subplots 
fig, axes = plt.subplots(2, 2, figsize=(10, 10)) 
# Original image 
axes[0, 0].imshow(image, cmap='gray') 
axes[0, 0].set_title('Original Image') 
axes[0, 0].axis('on') 
# Gaussian Blurred image 
axes[0, 1].imshow(gaussian_blur, cmap='gray') 
axes[0, 1].set_title('Gaussian Blur - Noise Reduction') 
axes[0, 1].axis('on') 
# Gradient Magnitude (with annotations) 
axes[1, 0].imshow(gradient_magnitude, cmap='gray') 
axes[1, 0].set_title('Gradient Magnitude & Direction') 
# Add arrows to indicate gradient direction 
for i in range(0, image.shape[0], 20): 
    for j in range(0, image.shape[1], 20): 
        direction = gradient_direction[i, j] 
        dx = np.cos(np.deg2rad(direction)) * 10  # Arrow length proportional to direction 
        dy = np.sin(np.deg2rad(direction)) * 10 
        axes[1, 0].arrow(j, i, dx, dy, head_width=3, head_length=3, fc='red', ec='red') 
        axes[1, 0].axis('on') 
        # Final edges detected after Canny 
        axes[1, 1].imshow(edges, cmap='gray') 
        axes[1, 1].set_title('Final Edges (Canny Detection)') 
        axes[1, 1].axis('on') 
        # Adjust layout for better visualization 
        plt.tight_layout() 
        plt.show()
```
