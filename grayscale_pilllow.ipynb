```python
# Pillow: The convert('L') method is used to convert the image to grayscale, where 'L' stands for "luminance," which represents the grayscale mode. 
from PIL import Image 
# Load an image 
image = Image.open('D:\\anaconda\\images\\img2.jpg') 
# Convert the image to grayscale 
gray_image = image.convert('L') 
# Save the grayscale image 
gray_image.save('gray_image_pillow.jpg') 
# Display the grayscale image 
gray_image.show() 
```
